```python
from sklearn.svm import SVC
from sklearn import datasets
from sklearn.model_selection import train_test_split
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report, f1_score
```

```python
pd = datasets.load_iris(as_frame=True).frame
```

```python
pd.isnull().sum()
```

```python
X = pd.drop("target", axis=1)
y = pd["target"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

```python
scaler = StandardScaler()
X_scaled_train = scaler.fit_transform(X_train)
X_scaled_test = scaler.transform(X_test)
```

```python
# by default = RBF (Gaussian) kernel
```

```python
model = SVC()
model.fit(X_scaled_train, y_train)
y_pre = model.predict(X_scaled_test)
```

```python
print("Accuraccy will be :", accuracy_score( y_test, y_pre))
```

```python
print("Accuraccy will be : \n ", classification_report( y_test, y_pre))
```

```python
# linear 
```

```python
model = SVC(kernel="linear")
model.fit(X_scaled_train, y_train)
y_pre = model.predict(X_scaled_test)
print("Accuraccy will be :", accuracy_score( y_test, y_pre))
```

```python
# polynomial
```

```python
model = SVC(kernel="poly")
model.fit(X_scaled_train, y_train)
y_pre = model.predict(X_scaled_test)
print("Accuraccy will be :", accuracy_score( y_test, y_pre))
```

```python
# sigmoid 
```

```python
model = SVC(kernel="sigmoid")
model.fit(X_scaled_train, y_train)
y_pre = model.predict(X_scaled_test)
print("Accuraccy will be :", accuracy_score( y_test, y_pre))
```
